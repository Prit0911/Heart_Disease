```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
df = pd.read_csv('heart_disease.csv')
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df.info()
```

```python
X = df.drop('Heart Disease Status',axis=1)
y = df['Heart Disease Status']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=42,stratify=y)
```

```python
model = RandomForestClassifier(random_state=42)
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print(y_pred)
```

```python
print("Accuaracy:", accuracy_score(y_test,y_pred))
```

```python
print("Classification Report:",classification_report(y_test,y_pred))
print("\n Confusion Matrix:", confusion_matrix(y_test,model.predict(X_test)))
```

```python
def predict_from_input(input_data):
    # Convert dict to DataFrame with same column order as training set
    input_df = pd.DataFrame([input_data], columns=X.columns)
    prediction = model.predict(input_df)[0]
    return "Heart Disease" if prediction == 1 else "No Heart Disease"

sample_input = {
    "age": 62,
    "sex": 1,
    "cp": 3,
    "trestbps": 150,
    "chol": 260,
    "fbs": 1,
    "restecg": 0,
    "thalach": 120,
    "exang": 1,
    "oldpeak": 3.5,
    "slope": 0,
    "ca": 2,
    "thal": 2
}

print("Prediction:", predict_from_input(sample_input))
```

```python
preds = model.predict(X_test)
print(set(preds))
print(confusion_matrix(y_test, preds))
print(classification_report(y_test, preds))
```
